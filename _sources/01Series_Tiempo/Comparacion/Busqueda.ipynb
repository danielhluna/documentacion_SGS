# Comparación de imágenes descargadas vs ssara_listing.txt

Este script se creo con la finalidad, de al momento de hacer un `split_jobs.py run` no se encuentren todas las imágenes en el repositorio de ASF, se pueda saber cuales son las faltantes y las fechas que se tienen que excluir de la serie de tiempo.

Para esto se tienen que crear dos archivos o descargar el archivo `ssara_listing.txt`, se describirá la forma en que lo hice, se realiza un cat del ssara_listing.txt y este mostrara toda la información del archivo, se copia de la consola y se pega en un archivo de texto plano, se guardo este documento como ssara_list.txt

```bash
cat ssara_listing.txt
```

Para ver que imágenes se descargo, realizamos el siguiente comando `ls -l *.zip`, copiamos el resultado lo pegamos en una hoja de calculo, separamos el texto por columnas, se le indica que el espacio el es el fin de columna, por lo que separa la información, nos quedamos con la columna de nombres de las imágenes con terminación `.zip` y pegamos esa columna en un archivo de texto plano, y lo guardas con el nombre de `lista.txt`

```bash
ls -l *.zip
```

Teniendo los archivos ejecuta los siguientes comandos

````python
# Archivo que contiene las rutas
ssara_list="ssara_list2.txt"

# Archivo que contiene los nombres
lista="lista2.txt"

# Verificar si los archivos existen
if [ ! -f "$ssara_list" ]; then
  echo "El archivo de rutas '$ssara_list' no existe."
  exit 1
fi

if [ ! -f "$lista" ]; then
  echo "El archivo de nombres '$lista' no existe."
  exit 1
fi

# Recorrer las rutas y verificar si los nombres están en la lista
while IFS= read -r ruta; do
  # Extraer el nombre de archivo de la ruta
  nombre_archivo=$(basename "$ruta")

  # Verificar si el nombre de archivo no está en la lista
  if ! grep -q "$nombre_archivo" "$lista"; then
    echo "Archivo no encontrado en la lista: $nombre_archivo"
  fi
done < "$ssara_list"
````

*Se despliega el nombre de los archivos que no se encuentran en la lista de imágenes descargadas*

En el template se pueden descartar estas fechas, para la serie de tiempo.
