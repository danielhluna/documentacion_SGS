# Evaluación de la Vulnerabilidad Social

_________

Herramienta para la evaluación de la vulnerabilidad socioeconomica en zonas urbanas.

Con el fin de evaluar la vulnerabilidad socioeconomica se utiliza los censos de población del 2010 o 2020 realizados por el INEGI. La información esta separada en las tablas donde se describe factores socioeconómicos y la información vectorial que contiene en donde se levantaron dichos datos. 

El censo trata de recabar las principales características de la población y conocer su distribución en el territorio nacional, los censos en México se realzan cada 10 años, tiene características que lo hacen compatible con censos en otros países. El Censo 2020 aumenta el número de preguntas y pasa de 29 en 2010 en el Cuestionario Básico a 38 y de 75 a 103 en el Ampliado.

El siguiente enlace contiene una descripción detallada del [censo de población](https://www.inegi.org.mx/contenidos/productos/prod_serv/contenidos/espanol/bvinegi/productos/nueva_estruc/702825197520.pdf).

## Información Geográfica

Para este análisis de vulnerabilidad, usaremos información espacial que esta relacionada con la información de la base de datos anterior, dependiendo del censo se tiene que relacionar con la información correspondiente:

Para el censo 2020 se utiliza el [Marco Geoestadístico. Censo de Población y Vivienda 2020](https://www.inegi.org.mx/app/biblioteca/ficha.html?upc=889463807469).

Para el censo 2010 se usa el [Marco Geoestadístico. Censo de Población y Vivienda 2010 version 5](https://www.inegi.org.mx/app/biblioteca/ficha.html?upc=702825292812).

El Marco Geoestadístico (MG) es un producto que integra información vectorial, tablas de atributos y catálogos. Muestra la división geoestadística del territorio nacional en sucesivos niveles de desagregación. Esta división está dada por los llamados LÍMITES GEOESTADÍSTICOS, que pueden coincidir con los límites político-administrativos oficiales, los cuales tienen sustento legal; sin embargo, los que no cuentan con dicho sustento deben entenderse como límites provisionales, trazados sólo para realizar los operativos censales.

````r
capa_ageb <- read_sf(dir_shp)
capa_ageb <- st_transform(capa_ageb, "+proj=longlat +datum=WGS84 +no_defs")
````

````r
capa_ageb$area_km2 <- st_area(capa_ageb) / 10^6
#convertir el tipo de columna
capa_ageb$area_km2 <- as.numeric(capa_ageb$area_km2)
````

````r
# Coordenadas del centro de la zona de interes, cambiar
latitud <- 22.155152
longitud <- -100.974473
zoom <- 11  # Nivel de zoom deseado
````

````r

# Crear una paleta de colores para representar la población de las AGEB
pal <- colorQuantile("Blues", domain = capa_ageb$area_km2, n=5)

# Crear un mapa interactivo con Leaflet
mapa <- leaflet(capa_ageb) %>%
  addTiles() %>%
  addPolygons(
    fillColor =  ~pal(area_km2),
    fillOpacity = 0.5,
    color = "black",
    weight = 1,
    popup = ~paste("Área:", round(area_km2,3), "km²"), # Muestra los datos al hacer clic en un polígono
    label = ~CVEGEO # Indica el nombre al poner el cursor
  ) %>%
  addLegend(
    pal = pal,
    values = ~area_km2,
    title = "Área en kilómetros cuadrados",
    opacity = 1,
    position = "bottomright"  # Cambia la posición de la leyenda si es necesario
  ) %>%
  setView(lng = longitud, lat = latitud, zoom = zoom)  # Establece el centro y el nivel de zoom

# Establecer la altura del mapa en 600px
mapa <- htmlwidgets::onRender(mapa, '
  function(el, x) {
    $(el).css("height", "800px");
  }
')

# Agregar un título utilizando etiquetas HTML en la leyenda
mapa <- mapa %>% 
  addControl(
    html = '<div style="text-align: center; background-color: #fff; padding: 10px;"><h3>Mapa de las AGEB de San Luis</h3></div>',
    position = "topright"  # Cambia la posición del título si es necesario
  )

# Mostrar el mapa interactivo
mapa
````

Función para escalar los indices para la evaluación de la vulnerabilidad socioeconomica, utilizando el metodo de normalización mínimo-máximo

````r
nor_mm <- function(x) {
  rng <- range(x, na.rm = TRUE, finite = TRUE)
  (x - rng[1]) / (rng[2] - rng[1])
}
````
